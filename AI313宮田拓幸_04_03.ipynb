##Geminiのchainの実装例

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_google_genai import ChatGoogleGenerativeAI

#プロンプトテンプレートを生成
prompt = ChatPromptTemplate.from_messages([
    ("system","ユーザーが入力した料理のレシピを考えてください。"),("human","{dish}")
])
#モデルのインスタンスを生成
model = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash",google_api_key=GOOGLE_API_KEY,temperature=0.7
)
```

##Chainの流れを作り、実行する

```python
#LCELで流れを作る
chain = prompt | model
```

```python
#chainを実行してはじめて、モデルの呼び出しが行われる
ai_message = chain.invoke( {"dish": "カレー"} )
print( ai_message.content)
```

```python
print( ai_message.content[0]["text"])
```

 StrOutputParserを流れに加える

```python
from langchain_core.output_parsers import StrOutputParser
#既存のcainに追加する
str_chain = chain | StrOutputParser()
# 3つのインスタンスをつなげる例
str_3_chain = prompt | model | StrOutputParser()
```

```python
#実行すると、modelの応答のAIMessageが文字列に変換される
output = str_chain.invoke({"dish":"チャーハン"})
print( output )
```

```python
#実行すると、modelの応答のAIMessageが文字列に変換される
output = str_3_chain.invoke({"dish":"シチュー"})
print( output )
```

##Pydanticによるオリジナルのデータ型で入出力制御する

```python
from pydantic import BaseModel, Field
from typing import List

#出力形式をPydanticで定義
class Recipe(BaseModel):
  ingredients: List[str] = Field(description="料理の材料リスト")
  steps: List[str] = Field(description="料理手順のリスト")
#class定義おわり
```

##定義したRecpieクラスをモデルのインスタンスの出力に設定済みにする

```python
#with_structured_putメソッドを使って出力形式をRecpie型にする
structured_model = model.with_structured_output(Recipe)
#promptとチェーンをつなげる
chain = prompt | structured_model

#chainを実行
recipe = chain.invoke({"dish":"サムギョプサル"})

#typeで型を表示
print(type(recipe))
#生で出力
print(recipe)
```

##PydnaticPutputParserでRacipe型の形式をプロンプトに設定

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import PydanticOutputParser

#Recipe型を解析するOutPutParserを定義
output_parser = PydanticOutputParser(pydantic_object=Recipe)
```

```python
print(output_parser.get_format_instructions())
```

##プロンプトテンプレートを生成


```python
#プロンプトテンプレートを生成
prompt_template = ChatPromptTemplate.from_messages([
    ("system","ユーザーが入力した料理のレシピを考えてください。\n\n{format_instruction}"),("human","{dish}")

])
#フォーマット付きのプロンプト文にする
prompt_with_format = prompt_template.partial(format_instruction=output_parser.get_format_instructions())
```

```python
#prompt_with_formatを使ったチェーンを作る
chain = prompt_with_format | model | output_parser
#チェーン実行
recipe = chain.invoke({"dish":"サムギョプサル"})
print(type(recipe))
print(recipe)
```
